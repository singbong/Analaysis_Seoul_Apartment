```python
from xgboost import XGBRegressor
from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.metrics import mean_squared_error
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
plt.rcParams['font.family'] = 'NanumGothic'
plt.rcParams['axes.unicode_minus'] = False
import warnings
warnings.filterwarnings(action='ignore')



df_anc= pd.read_feather("../data/make_data/df_anc_for_model.feather")
df= df_anc.copy()
df.drop(columns=['아파트명'],inplace=True)
```

```python
df.select_dtypes(include='category').columns
```

```python
df = pd.get_dummies(df, columns=['호선'], prefix='호선').copy()
df = pd.get_dummies(df, columns=['건물구조'], prefix='건물구조').copy()
df = pd.get_dummies(df, columns=['자치구명'], prefix='자치구명').copy()

target_feature = 'target'

# target feature를 맨 뒤로 이동
columns = [col for col in df.columns if col != target_feature] + [target_feature]
df = df[columns]
```

```python
df= df_anc.copy()
df.drop(columns=['아파트명'],inplace=True)
df = pd.get_dummies(df, columns=['호선'], prefix='호선').copy()
df = pd.get_dummies(df, columns=['건물구조'], prefix='건물구조').copy()


target_feature = 'target'

# target feature를 맨 뒤로 이동
columns = [col for col in df.columns if col != target_feature] + [target_feature]
df = df[columns]
df.info()
```

```python
def local_train(local):
    print(f"\n=== {local} 분석 결과 ===")
    
    df_local = df[df['자치구명']==local].copy()
    df_local.drop(columns='자치구명',inplace=True)
    
    model = XGBRegressor(
        tree_method='hist',
        device='cuda',
        n_estimators=3000,
        random_state=42,
        enable_categorical=True
    )

    X_train, X_test, Y_train, Y_test = train_test_split(df_local.iloc[:,:-1], df_local.iloc[:,-1], test_size=0.2, random_state=42)
    X_tr, X_val, Y_tr, Y_val = train_test_split(X_train, Y_train, test_size=0.2, random_state=42)

    model.fit(X_tr, Y_tr, eval_set=[(X_tr, Y_tr),(X_val, Y_val)], eval_metric='rmse', early_stopping_rounds=300, verbose=False)
    rmse = mean_squared_error(Y_test, model.predict(X_test), squared=False)
    print(f"RMSE: {rmse:.4f}")

    # SHAP 값 계산
    explainer = shap.TreeExplainer(model)
    shap_values = explainer.shap_values(X_test)
    
    feature_importance = np.abs(shap_values).mean(0)
    feature_importance_df = pd.DataFrame({
        'feature': X_test.columns,
        'importance': feature_importance
    })
    
    top_20_features = feature_importance_df.sort_values('importance', ascending=False).head(20)
    features = top_20_features['feature'].tolist()

    # 부분 의존성 플롯 생성
    plt.figure(figsize=(20, 15))
    fig, ax = plt.subplots(4, 5, figsize=(10, 10))
    display = PartialDependenceDisplay.from_estimator(
        model,
        X_test,
        features,
        kind="average",
        subsample=1000,
        n_jobs=-1,
        grid_resolution=20,
        random_state=42,
        ax=ax
    )
    fig.suptitle(f"{local} - Partial Dependence Plots", fontsize=16)
    plt.tight_layout()
    plt.show()

# 모든 자치구에 대해 실행
unique_locals = df['자치구명'].value_counts().index
for local in unique_locals:
    local_train(local)
```
